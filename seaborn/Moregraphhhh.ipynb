```python
import seaborn as sns 
import matplotlib.pyplot as plt 
import seaborn as sns 
```

```python
df=sns.load_dataset('tips')
print(df)
```

```python
sns.histplot(x = "total_bill" , data = df , bins = 5 , color = "red" , hue = "sex", multiple = "stack" )
plt.show()
```

```python
sns.histplot(x = "total_bill" , data = df , bins = 5 , color = "red" , hue = "sex" , multiple = "layer" )
plt.show()
```

```python
sns.histplot(x = "total_bill", data = df , bins = 5 , color = "red" , hue = "sex" , multiple = "layer" , element = "bars" , kde = True , palette = "Blues" )
plt.grid()
plt.title("HistoGramChart")
plt.show()
```

```python
sns.kdeplot(x= "total_bill" , data = df, hue =  "sex" , bw_adjust = 2 , clip = (0,25) , fill=True , multiple = "layer")
plt.show()
```

```python
sns.kdeplot(x = "tip" , data = df , hue = "day" , bw_adjust = 2 , clip = (0 ,25) , fill = False , multiple = "layer" ,hue_order =[ "Sun" , "Sat" , "Fri" , "Thur" ])
plt.show()
```

```python
x = [10,20,30,40,50,60,70,80,90]

sns.boxplot(x = x , color = "green"  , saturation = 1, orient = "y")
plt.show()
```

```python
sns.boxplot(x = "total_bill" , data = df , hue = "sex" , color = "red" , saturation = 0.5 , whis = 3 , linecolor = "y" , linewidth = 2)
plt.show()
```

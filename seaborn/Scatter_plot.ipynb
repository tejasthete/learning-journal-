```python
import pandas as pd
import matplotlib.pyplot as plt 
import seaborn as sns
```

```python
df =  pd.DataFrame({"SHour":[1,2,3,4,5] , "Marks":[20,40,60,80,100]})

sns.scatterplot(x="SHour" , y = "Marks" , data = df , hue = "Marks" , 
                size = "Marks" , sizes = (50 ,200), color =" coolwarm")
plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.title("Study Hours vs Marks")
plt.show()
# x, y --------> the columns to plot on each axis
# data --------> the dataframe containing x and y
# hue ---------> colors the points based on a column's values
# size --------> scales point size based on a column's values
# sizes -------> (min, max) range for point sizes
# palette -----> color scheme for hue
```

```python
name = ["A" , "B" , "C" ,"D" ]
marks = [35,67,89,90]
gender = ["M" , "F" , "M" ,"F"]
s  = {"Name":name ,"Marks":marks , "Gender":gender}
df= pd.DataFrame(s)
print(df)
```

```python
#Syntax for line plot --> sns.lineplot(x=value , y = value , data = DataFrame_name)
sns.lineplot(x="Name" , y="Marks" , data = df , 
             hue = "Gender" , style="Gender" , 
             marker= "o", markersize = 10 , 
             linewidth = 2 , linestyle="dotted",
             legend=False
              )
#hue---> makes the group 
#style and linesyle is similar but use one of them
#automatically lengend is highlited to hide legend we have to use 
plt.show()
```

```python
import pandas as pd
import seaborn as sns

d = pd.read_csv("tipss.csv")

sns.lineplot(x="total_bill", y="tip", data=d ,hue = "sex" ,style='sex', hue_order=["Female" , "Male"],
             markers =["o","*"],markersize = 10 ,dashes=False , palette = "muted" , )
plt.show()
```

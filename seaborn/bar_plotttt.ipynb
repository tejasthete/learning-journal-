```python
import pandas as pd
```

```python
df = pd.DataFrame({"Dept": ["IT" , "HR" , "Analyist" , "IT" , "HR"] ,
                   "Revenue" : [ 10000,20000,40000,11003,13402] , 
                   "Gender" : [ "Female" , "Male" , "Male" , "Female" , "Female"]})
print(df)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
sns.barplot(x = "Dept" , y = "Revenue" , data =df , errorbar = None ,hue = "Gender" ,
            order = ["HR" , "Analyist" , "IT"],
            hue_order = ["Male" , "Female"], palette='dark:red',
            estimator= 'mean'  , width = 0.8 , 
           )
plt.show()


#errorbar= None ---> for removing the error bar
#hue ---> for gruping 
#dodge---> for seprating the data
#order---> for changing the position of the data

```

```python
import pandas as pd

d = pd.read_csv("tipss.csv")

d.head()
```

```python
sns.barplot(x = "day" , y = "total_bill" , data = d , 
            errorbar = None ,
            hue = "sex" , 
            dodge =True , 
            width =0.3 , 
            palette = "muted")
plt.show()
```

```python
sns.barplot(x = "smoker" , y = "tip" , data = d , 
            errorbar = None ,
            hue = "sex" , 
            dodge =True , 
            width =0.3 , 
            palette = "Blues",
           order = ["Yes" , "No"],
            hue_order = ["Male" , "Female"]
            
           )

plt.show()
```

```python
sns.barplot(x = "time" , y = "total_bill" , data = d , 
            hue = "sex" , 
            dodge =True, 
            width =0.3 , 
            palette = "Greens",
            hue_order = [ "Male" , "Female" ], 
            capsize = 20,
            estimator = 'sum' 
           )

plt.show()
```

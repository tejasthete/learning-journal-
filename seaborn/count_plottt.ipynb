```python
import pandas as pd
import seaborn as sns 
import matplotlib.pyplot as plt

d = pd.read_csv("tipss.csv")

d.head()
```

```python
sns.countplot(x = "sex"  , data = d , hue = "sex" , width = 0.3 , palette = 'pink' )
plt.show()
```

```python
sns.countplot(x = 'day' , data = d , hue = 'sex' , palette = 'Blues' , width = 0.5,
              legend =True
    )
plt.show()
```

```python
sns.pairplot(data = d , hue = "sex")
plt.show()
```

```python
sns.pairplot(data = d ,vars= ['total_bill' , 'size'], hue = "sex")
plt.show()
```

```python
sns.pairplot(data = d , x_vars= ['total_bill' , 'size'], hue = "sex")
plt.show()
```

```python
sns.pairplot(data = d , y_vars= ['total_bill' , 'size'],  hue = "sex")
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns 
df = pd.DataFrame({"SHour":[1,2,3,4,5],"Marks":[20,40,60,80,100]})
print(df)
```

```python
corr = df["SHour"].corr(df["Marks"])
print(corr)
```

```python
sns.heatmap(df)    #always works on 2D
plt.show()
```

```python
import numpy as np 
df = np.array([[90 , 55 ,70 ],
               [45 , 10 , 40],
               [30 , 65 , 80]])
print(df)
```

```python
sns.heatmap(df ,annot = True , fmt = '1f' , linewidths = 1 , linecolor = "yellow" , 
           annot_kws ={"size": 10 , "color":"red"} , cmap = "coolwarm" , cbar = True , xticklabels = ["Tejas" , "Rahi" , "Bhavana"] 
           , yticklabels = ["Maths","Sql","py"] , cbar_kws ={"label": "Data" })
plt.xlabel("Student Name")
plt.ylabel("Subject Name")
plt.title("Score card Report")

plt.show()
#annot ------>for adding the number inside the chart 
#format/fmt-->"d".2f" ----> for deciaml number---->2f for 2 decimal number 
#linwidth ---> for showing the gap of the between the box
#linecolor---> for the color of the color between boxex 
#annot_kws---> for the size of th number inside the box  or changing the color inside the box 
#cmap -------> for changing the colors of the heatmap ( Reds , Organges , Purple , YlOrRd , coolwarm , RdBu , viridis , Magma) 
#cbar -------> for displaying the color bar 
#xticks_labels----> for labeling the x axis
#yticks_labels----> for labeling the y-axis
#vmin----> for staring with minimum number
#vmax----> for starting with maximum number
#cbar_kws----->for highlighting seabar
```

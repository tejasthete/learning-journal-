```python
import pandas as pd 
import numpy as np 
```

```python
df = pd.read_excel("../data/Pizza_Sales_Dataset.xlsx")
```

```python
df.head()
```

```python
#Data Features 
df["Date"] = pd.to_datetime(df["Date"])
df["Year"] = df["Date"].dt.year
df["Month"] = df["Date"].dt.month
df["Month_Name"] = df["Date"].dt.month_name()
df["Day"] = df["Date"].dt.day
df["Day_Name"] = df["Date"].dt.day_name()
```

```python
df.head()
```

```python
#Revenue Per Pizza
df["Revenue_Per_Pizza"] = df["Revenue"]/df["Sales"]
```

```python
df[["Sales" , "Revenue" , "Revenue_Per_Pizza"]].head()
```

```python
#Unsold Rates
df["Unsold_Rate"] = (df["Unsold"] / (df["Sales"] + df["Unsold"]) )*100
```

```python
df[["Sales" ,"Unsold" , "Unsold_Rate"]].head()
```

```python
#Total Production

df["Total_Production"] = df["Unsold"] + df["Sales"]
```

```python
df[["Sales" , "Unsold" , "Total_Production"]].head()
```

```python
#Sales Rate
df["Sales_Rate"] = (
    df["Sales"] / df["Total_Production"] * 100 )
```

```python
df[["Sales" , "Unsold" , "Total_Production" , "Sales_Rate"]].head()
```

```python
df["Sales_Performance"] = pd.qcut(df["Sales"], q = 3, labels = ["Low" , "Medium" , "High"])
```

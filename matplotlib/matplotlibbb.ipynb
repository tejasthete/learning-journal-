```python
import matplotlib.pyplot as plt
import numpy as np 
```

```python
x = np.arange(0,11,2)
print(x)
```

```python
y = x**2
print(y)
```

```python
z  = y**2
print(z)
```

```python
plt.plot(x,y,color="red",
         ls = "solid" , 
         marker= "o",
         ms = 15 , 
         mec="y",
         mfc = "c",
         lw = 5,
         label = "Trends_Gowth" , 
         mew=5)
plt.xlabel("Xaxis",fontsize = 20 , color = "red")
plt.ylabel("Yaxis",fontsize = 20 , color = "orange")
plt.title("LineChart" , fontsize=20 , color = "cyan")
plt.legend(loc="upper left",
           fontsize = 15,
           title = "Line" , title_fontsize  = 5,
           frameon = True , 
           facecolor = "yellow" ,
           shadow = True,
           edgecolor = "red" , 
           fancybox = False,
           framealpha =1 )
plt.grid(axis = "both",color="y" , ls = "dotted" , lw = 3)
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
sub=["python","sql","java","manual"]
boys=[50,35,40,60]
girls=[25,10,70,30]
girls_Boys=[75,45,110,90]
width=0.3
x=np.arange(len(sub))
print(x) #[0 1 2 3]
y=[i+width for i in x]
print(y)
z=[i+width for i in y]
print(z)
plt.bar(x,boys,color="r",width=width,edgecolor="b",label="Boys")
plt.bar(y,girls,color="y",width=width,edgecolor="b",label="girls")
plt.bar(z,girls_Boys,color="c",width=width,edgecolor="b",label="Boys+Girls")
plt.xticks(x+width/2,sub)
plt.grid(color="black")
plt.legend()
plt.title("BARCHART",fontsize=20,color="r")
plt.show()

 
```

```python
import matplotlib.pyplot as plt
act = ["Game" , "food" , "study" , "sleeping"]
Rank = [25 , 70 , 15 , 90 ] 
plt.pie(Rank ,labels = act , explode = [0.1,0.0 , 0.0, 0.0],
        autopct = "%1.0f%%" ,colors = ["r" , "b" , "c" , "g" ] ,
        wedgeprops = {"linewidth":3 , 
                      "edgecolor": "y"})
plt.legend()
plt.show()
```

```python
#pie chart parameter 

# 1) var_name--->specifies the values of each slice 

# 2) labels----> Adds names to each slice

#3) autopct ---> Displays percentages on slices
    # 1 decimal place ---> %1.1f%%
    # No  decimal Place---> %1.0F%%
    # 2 decimal  places---> %!.2


#4) color ---> Changes slice color

#5) explode--->Separate slices from the center

#6)shadow ---> Add a showdow effect, show = True/False


#7) startangle--> Rotates the starting postion of the pie chart (0/90/180/120)

#8) counterclock---> Control direction of slices
    # counterclock = True  #Anticlockwise
    # counterclock = False #Clockwise

#9) radius----> Controls pie size(0.5/1.4/1.8)

#10)labeldistance----> Controls label distance from the center 

#11)wedgeprops---> Customize slice borders
#syntax---> wedgeprops = {edgecolor = 'color' , linewidth = num}

#12)textprops---> Customize label text
# syntax--->textprops = { 'fontsize':number , 'color':color})

#13) rotatelabel---> Rotates labeel accordng to slice angels 

```

```python
import matplotlib.pyplot as plt 
import numpy as np
x = np.arange(0,5,1)
print(x)
y = x**2
plt.scatter(x,y,marker = "*" , s =  500 , color = ["r" , "c" , "b" , "g" , "y"] , 
            edgecolor = "r",linewidth = 2 , label = "two-variable")
plt.xlabel("x-axis" , fontsize = 25 , color = "r")
plt.ylabel("y-axis" , fontsize = 25 , color = "b")
plt.title("Scatter Plot" , fontsize = 20 , color = "g")
plt.grid()
plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt
x = [1,2,3,4,5]
y = [10,15,20,25,30]
z = [15,20,25,30,35]
z1 = [20,25,30,35,40]
plt.figure(figsize=(8,8))
plt.suptitle("SubPLOTCharts")
plt.subplot(2,2,1)
plt.plot(x,y,ls = "dotted" ,marker = "*")
plt.title("First plot")
plt.grid()

plt.subplot(2,2,2)
plt.scatter(y,z,s = 400 ,marker = "+" ,color = "r")
plt.title("Second Plot")

plt.subplot(2,2,3)
plt.hist(z1, color = "r" )
plt.title("Third Plot")

plt.subplot(2,2,4)
plt.pie(z,labels = ["a" , "b" , "c" , "d" , "e"])
plt.title("Fourth Plot")
plt.legend()



plt.show()
```

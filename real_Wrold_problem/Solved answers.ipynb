```python
# 1. Total Bookings by Hotel (Count Plot)  

sns.countplot(x='hotel' , data = df , color = "green" , width = 0.3 , edgecolor = 'black')  # if i use y axis it will change vertically
plt.title("Total Booking By Hotel")
plt.xlabel("Hotels")
plt.ylabel("Count")
```

```python
# 2.What percentage of bookings were cancelled?  pie chart
```

```python
cancelled = df["is_canceled"].value_counts()



plt.pie(
    cancelled,
    labels=["Not Cancelled", "Cancelled"],
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Percentage of Bookings Cancelled")
plt.show()
```

```python
# 3. Which Hotel Has More Cancelled Bookings? — Stacked Column

hotel_cancel = df.groupby(["hotel", "is_canceled"]).size().unstack()

hotel_cancel.plot(kind="bar", stacked=True)

plt.title("Cancelled vs Not Cancelled Bookings by Hotel")
plt.xlabel("Hotel")
plt.ylabel("Number of Bookings")
plt.show()
```

```python
# 4.Which market segment has the highest number of cancelled bookings? Bar chart


kj = df[df['is_canceled']==1].groupby("market_segment").size().sort_values(ascending = False)

plt.figure(figsize = (10,5))

sns.barplot(
    x = kj.index,
    y = kj.values, palette = "coolwarm" , hue = kj.index , legend = True)



plt.xlabel = ("Market_Segment")
plt.ylabel = ("Booking Cancellation")
plt.show()
```

```python
# 5. Which Hotel Has Higher Average Room Rate (ADR)? — Column Chart

avg_adr = df.groupby("hotel")["adr"].mean()

plt.figure(figsize=(8, 5))

sns.barplot(
    x=avg_adr.index,
    y=avg_adr.values,
    color = "green"
)

plt.xlabel = ("Hotel")
plt.ylabel = ("Avg ADR")
plt.show()
```

```python
# 6. Which Month Has Highest Booking Demand? — Line Chart



monthly_bookings = (
    df["arrival_date_month"]
    .value_counts()

)

plt.figure(figsize=(12, 5))

sns.lineplot(
    x=monthly_bookings.index,
    y=monthly_bookings.values,
    marker="o"
)

plt.title("Monthly Booking Demand")

plt.show()
```

```python
# 7. During Which Month Is ADR Highest? — Line Chart


monthly_adr = ( 
    df.groupby("arrival_date_month")["adr"].mean()
)

plt.figure(figsize = (12,5))


sns.lineplot(
    x = monthly_adr.index,
    y = monthly_adr.values,
    marker = "o"

)
plt.title("Average Daily Rate by Month") 

plt.show()
```

```python
# 8. Which Customer Type Makes the Most Bookings? — Donut Chart

customber_counts = df["customer_type"].value_counts()

plt.figure(figsize=(7,7))

plt.pie(customber_counts,
       labels = customber_counts.index , 
       autopct = "%1.1f%%",
       startangle = 90,
       wedgeprops = dict(width = 0.5))

plt.title("Booking by customber Type")           
plt.show()
```

```python
# 9. Relationship Between Lead Time and Cancellation — Scatter Chart

plt.figure(figsize=(8, 5))

sns.stripplot(
    x="is_canceled",
    y="lead_time",
    data=df,
    alpha=0.3
)

plt.title("Lead Time vs Cancellation")
plt.show()
```

```python
from scipy.cluster.hierarchy import linkage, dendrogram
import matplotlib.pyplot as plt
from sklearn.cluster import AgglomerativeClustering
```

```python
data=[[100],[120],[140],[500],[520],[540]]
customers=['a','b','c','d','e','f']
```

```python
model=AgglomerativeClustering(n_clusters=2,linkage='single')
pred=model.fit_predict(data)
```

```python
pred
```

```python
res=linkage(data,method='single')
dendrogram(res,labels=customers)
plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd
from scipy.cluster.hierarchy import linkage,dendrogram
from sklearn.cluster import AgglomerativeClustering
from sklearn.datasets import load_iris
```

```python
iris=load_iris()
x=iris.data
```

```python
model=AgglomerativeClustering(n_clusters=3,linkage='ward')
labels=model.fit_predict(x)
```

```python
df=pd.DataFrame(x,columns=iris.feature_names)
df
```

```python
df['cluster']=labels
```

```python
plt.scatter(x[:,0],x[:,1],c=labels)
```

```python
res=linkage(x,method='ward')
dendrogram(res,labels=labels)
plt.show()
```

```python
df=pd.DataFrame({'name':['darshan','on','oi','op','ew','dd','f','r'],'hours':[1,1.5,1.9,2,2.1,2.5,3,3.5]})
x=df[['hours']]
```

```python
model=AgglomerativeClustering(n_clusters=3,linkage='ward')
pred=model.fit_predict(x)
```

```python
df['clusters']=pred
```

```python
plt.figure(figsize=(10,10))
res=linkage(x,method='ward')
dendrogram(res,labels=df['name'].values)
plt.show()
```

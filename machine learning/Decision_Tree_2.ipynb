```python
import pandas as pd 
```

```python
data = pd.DataFrame({"Age" : [ 18,21,26,30,35],
        "FB_user_Min": [10,20,30,40,50],
        "clicked": ["NO" , "NO" , "YES" , "YES" , "NO" ]})

print(data)
```

```python
x = data[["Age" ,"FB_user_Min"]]
y = data["clicked"]
```

```python
from sklearn.model_selection import train_test_split

x_train , x_test , y_train ,y_test  = train_test_split(x,y,test_size = 0.2 , random_state = 42)
```

```python
from sklearn.tree import DecisionTreeClassifier

model = DecisionTreeClassifier(criterion = "gini" , random_state = 42 , max_depth = 3)
model.fit(x_train , y_train)
```

```python
prediction = model.predict(x_test)
print("Actual_data" , y_test)
print("Predicted_test" , prediction)
```

```python
from sklearn.tree import plot_tree
import matplotlib.pyplot as plt
plt.figsize = (13,9)

plot_tree(model , feature_names = x.columns, class_names =  model.classes_ ,rounded = True, filled = True)
plt.show()
```

```python
from sklearn.metrics import(accuracy_score,precision_score , recall_score , f1_score)
```

```python
ac = accuracy_score(y_test , prediction)
ps = precision_score(y_test, prediction)
rs = recall_score(y_test , prediction)
f1 = f1_score(y_test,prediction)

print("Accuracy Score------>",ac)
print("Precision Score----->",ps)
print("Recall Score-------->",rs)
print("F1 Score------------>",f1)
```

```python
from sklearn.metrics import classification_report
cr = classification_report(y_test , prediction)
print("Classification_Report\n" ,cr)
```

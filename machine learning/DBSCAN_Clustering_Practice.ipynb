```python
# DBSCAN---> density based spatial clustring of application with noise
#form of unsupervised machine learning 
# automatiocally group will be from and before that we dont know how many group will be form 
# we need to point the centre point----> for example there is the hurdle of player the center men is giving speech the cener is
#nothing but centre point 
# boundary point----> 
# noise---->point outside the boundary condition is called noise 
#parameter--->1) eps------>also call as threshold ----> formulla--->distance--->d = root of (x2-x1)2 + (y2-y1)
             #2)minisample---->how many group we are creating is called minisample 
from sklearn.cluster import DBSCAN
```

```python
import pandas as pd 
```

```python
data = {"Student_Name": ["A" , "B" , "C" , "D" , "E" , "F" , "G" , "H" ] , 
        "Study_Hours" : [ 1,2.5,3,4.5,5,6,7.8,8.5],
        "Marks":[25,40,45,50,55,78,75,90]}
```

```python
df = pd.DataFrame(data)
print(df)
```

```python
x = df[["Study_Hours" , "Marks"]]
```

```python
from sklearn.cluster import DBSCAN 
```

```python
model = DBSCAN(eps = 0.5 , min_samples=3)
```

```python
df["cluster"] =model.fit_predict(x)
```

```python
df
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10,8))

plt.scatter(df["Study_Hours"],df["Marks"],
           c=df["cluster"],s=100 ,cmap = "Paired"
           )
# for i,row in df.iterrows():
#     plt.annotate(row["Student_Name"],(row["Study_Hours"],row["Marks"]),
#                 xytext=(6,6) ,textcoords=  "offset points")
                                      

plt.title("Study_Hours vs MArks")
plt.xlabel("Study_Hours")
plt.ylabel("Marks")
plt.grid(True)
```

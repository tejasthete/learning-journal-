```python
import random
```

```python
#robot example
start_Point = 0
end_Point = 10

while start_Point != end_Point:
    action = random.choice(["left","right"])

    if action =="left":
        start_Point+=1

    else:
        start_Point = max(0,start_Point-1)

    print(action)
    print(start_Point)

    if start_Point == end_Point:
        print("Reward---->10")

    else:
        print("Penalty")
        print("Code Complete")
    
```

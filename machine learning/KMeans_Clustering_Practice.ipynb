```python
import pandas as pd 
```

```python
df=pd.read_csv('tipss.csv')
print(df)
```

```python
#1.Load the tips.csv dataset using Pandas and display the first 10 rows.
df.head(10)
```

```python
# 2.Display the number of rows and columns using shape.
df.shape
```

```python
# 5.Select total_bill and tip as features for clustering.

x = df[["total_bill" , "tip"]]
print(x)
```

```python
from sklearn.cluster import KMeans
```

```python
# 6.Create a K-Means model with n_clusters=3 and random_state=42.
model = KMeans(n_clusters = 3 , random_state = 42 )
```

```python
# 7.Train the model and assign cluster labels using fit_predict().
# 8.Add a new column named Cluster to the original DataFrame.
df["cluster"] = model.fit_predict(x)
print(df)
```

```python
# 9.Display the first 10 rows of the dataset with cluster labels.
df.head(10)
```

```python
# 10.Display the cluster centers using model.cluster_centers_.
print(model.cluster_centers_)
```

```python
# 12.Create a scatter plot showing each cluster in a different color.
# 13.Display the centroids using black X markers.
# 14.Add the chart title, axis labels, legend, and grid.
import matplotlib.pyplot as plt 

plt.figure(figsize=(10,6))
colors = ["Red","Yellow","Green"]

for i in range(3):
    group = df[df["cluster"] == i]
    
    plt.scatter(group["total_bill"],
                group["tip"],
                color=colors[i],s=100,
                label=f"Cluster {i}" )
    
    plt.scatter(model.cluster_centers_[:,0],model.cluster_centers_[:,1],s  = 200,color = "black",label = "CenterPoint",marker = 'x')

plt.xlabel("Total_Bill")
plt.ylabel("TIP")
plt.title("K-Means Clustering")
plt.legend()
plt.grid(True)
plt.show()
    
```

```python
# 1. Customer Segmentation
# Group customers by Annual Income (n lakhs) and Spending Score.
# Data: (2,20), (3,25), (2.5,22), (5,50), (6,55), (5.5,52), (8,80), (9,90),
# (8.5,85)
# 1.Create a Data Frame using the given data.
# 2.Select Annual Income and Spending Score as features.

Ann_income = [2,3,2.5,5,6,5.5,8,9,10] 
Spending_score = [20,25,22,50,55,52,80,90,95]

data = pd.DataFrame({ "Annual_Income" : Ann_income , "Spending_Score" : Spending_score })
print(data)
```

```python
x = data
print(x)
```

```python
# 3.Create a K-Means model with 3 clusters.
from sklearn.cluster import KMeans
Model = KMeans(n_clusters = 3 , random_state = 42)

# 4.Assign cluster labels to the customers.
data["cluster"] = Model.fit_predict(x)
print(data)
```

```python
print(Model.cluster_centers_)
```

```python
# 5.Display the final dataset and cluster centers.
# 6.Plot the clusters in different colors and centroids using black X markers.

import matplotlib.pyplot as plt 

plt.figure(figsize=(10,6))
colors = ["Red","Yellow","Green"]

for i in range(3):
    group = data[data["cluster"] == i]
    
    plt.scatter(group["Annual_Income"],
                group["Spending_Score"],
                color=colors[i],s=100,
                label=f"Cluster {i}" )
    
    plt.scatter(Model.cluster_centers_[:,0],Model.cluster_centers_[:,1],s  = 200,color = "black",label = "Cluster",marker = 'x')

plt.xlabel("Annual_Income")
plt.ylabel("Spending_Score")
plt.title("K-Means Clustering")
plt.legend()
plt.grid(True)
plt.show()
    
```

```python
# 2. Fitness Activity
# Group users by Walking Distance (km) and Calories Burned. Data:
# (1,100), (2,150), (1.5,120), (4,300), (5,350), (4.5,320), (7,500),
# (8,600), (7.5,550)

Distance = [1,2,1.5,4,5,4.5,7,8,7.5]
Calories = [100,150,120,300,350,320,500,600,500]

data = pd.DataFrame({ "Distance" : Distance , "Calories" : Calories })
print(data)
```

```python
# 1.Group users into 3 clusters
X = data
print(X)
```

```python
from sklearn.cluster import KMeans
MODEL = KMeans(n_clusters = 3 , random_state = 42)
```

```python
data["cluster"] = MODEL.fit_predict(X)
print(data)
```

```python
print(MODEL.cluster_centers_)
```

```python
import matplotlib.pyplot as plt 

plt.figure(figsize=(10,6))
colors = ["Violet","Green","Cyan"]

for i in range(3):
    group = data[data["cluster"] == i]
    
    plt.scatter(group["Distance"],
                group["Calories"],
                color=colors[i],s=100,
                label=f"Cluster {i}" )
    
    plt.scatter(MODEL.cluster_centers_[:,0],MODEL.cluster_centers_[:,1],s  = 200,color = "black",label = "Cluster",marker = 'x')

plt.xlabel("Distance")
plt.ylabel("Calories")
plt.title("K-Means Clustering")
plt.legend()
plt.grid(True)
plt.show()
```

```python
# Task3

# 3.Weather Patterns
# Group days by Temperature (°C) and Humidity (%). Data: (20,40),
# (22,45), (21,42), (28,60), (30,65), (29,62), (35,80), (37,85), (36,82).

Temperature = [20,22,21,28,30,29,35,37,36]
Humidity = [40,45,42,60,65,62,80,85,82]

DATA = pd.DataFrame({"Temperature":Temperature , "Humidity" :Humidity}) 

print(DATA)

```

```python
# 1.Apply K-Means with 3 clusters.

Y = DATA
print(Y)
```

```python
from sklearn.cluster import KMeans
MODEL = KMeans(n_clusters = 3 , random_state = 42)
```

```python
DATA["cluster"] = MODEL.fit_predict(Y)
print(DATA) 
```

```python
# 4.Create a visualization using different colors.

import matplotlib.pyplot as plt 

plt.figure(figsize=(10,6))
colors = ["Violet","Green","Red"]

for i in range(3):
    group = DATA[DATA["cluster"] == i]
    
    plt.scatter(group["Temperature"],
                group["Humidity"],
                color=colors[i],s=100,
                label=f"Cluster {i}" )
    
    plt.scatter(MODEL.cluster_centers_[:,0],MODEL.cluster_centers_[:,1],s  = 200,color = "black",label = "Cluster",marker = 'x')

plt.xlabel("Temperature")
plt.ylabel("Humidity")
plt.title("K-Means Clustering")
plt.legend()
plt.grid(True)
plt.show()
```

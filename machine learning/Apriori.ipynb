```python
#1) KMeans (done)
#2) DBScan (done)
#3) Apriori---->used for the bussiness pattern 
# EXAMPLE---> The standard example is market basket analysis: “customers who buy bread and butter also tend to buy milk.”
# customer      Product
#    A           Bread , Milk
#    B           Bread , Milk , Butter
#    C           Bread , Butter 
#    D           Milk , Butter
#    E           Bread , Milk , Butter
# you need to perform 3 operation in apriori
```

```python
#1) Support--->element frequently repeated / Total customer comming to the shop
#support(A) = Repeated element / Total Transcation
#support(Bread)  =     4/5 * 100----->80% 
#support(Milk)   =     4/5 * 100----->80% 
#support(Butter) =     4/5 * 100----->80%
#each produt selling percentage 

#but we need to calulate in once ---> 
#support (A,B) = pair of Repeated Element / Total Tanscation
#support(Bread,Milk)   = 3/5*100--->60%
#support(Millk,Butter) = 3/5*100--->60%
#support(Bread,Butter) = 3/5*100--->60

#support(A,B,C) = Repeated Element / Total Tanscation
#support(Bread , Milk , Butter)  = 2/5*100 = 40%

#if the pair of element is less than 60% then remove the pair
# support >= min_support
```

```python
#2) Confindence
# if i purchsing two two element together then fist elment how many times it is repeated 
# confidence(A,B) =  Both elment frequence / element(A)

# Bread and Milk ---> 3/4*100--->75%
# Milk and Butter---> 3/4*100--->75%
# Butter and Bread--> 3/4*100--->75%
```

```python
#3) Lift-----> Confidence(A,B) / Support(B)
#                Bread , Milk / support(Milk) ----> 0.75/0.8
#                Milk , Butter / support(Butter)--> 0.75/0.8
#                Butter , Bread/ support(Bread)---> 0.75/0.8
# the lift value is 0.975

# Lift > 1: positive association (bought together more than by chance)
# Lift = 1: independent
# Lift < 1: negative association (bought together less than by chance)
```

```python
import pandas as pd 
data1 = {"Laptop":[1,1,1,0,1,1],
         "Bag" : [1,1,1,1,1,1],
         "Charger" : [0,0,1,1,0,1]
        }
df1 = pd.DataFrame(data1)
df1
```

```python
%pip install mlxtend
```

```python
from mlxtend.frequent_patterns import apriori, association_rules
```

```python
device = apriori(df1,min_support = 0.5 , use_colnames = True)
device
```

```python
rules = association_rules(device , metric = 'confidence' , min_threshold   = 0.5)
print(rules[['antecedents' , 'consequents' , 'support' , 'confidence' , 'lift']])
```

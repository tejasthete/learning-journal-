```python
import pandas as pd 


data = pd.DataFrame({"Salary" :[20000,25000,30000,35000,40000,45000],
                    "Credit_Score":[500,600,650,700,750,800],
                    "Loan_apporoved":["NO" , "NO" , "NO","YES" , "YES" , "YES"]})

print(data)
```

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
model = DecisionTreeClassifier()
```

```python
x = data[["Salary" , "Credit_Score"]]
print(x)
```

```python
y = data[["Loan_apporoved"]]
print(y)
```

```python
#Gini impurity = the probability of misclassifying a randomly picked sample from the node,
       #if you assigned it a random label according to the node's own class proportions.
```

```python
model = DecisionTreeClassifier(criterion = "gini" ,
                               max_depth=2,
                               random_state = 42)
```

```python
model.fit(x,y)
```

```python
prediction = model.predict([[35000,700]])
print(prediction)
```

```python
new_prediction = model.predict([[50000,850]])
print(new_prediction)
```

```python
from sklearn.tree import plot_tree
import matplotlib.pyplot as plt 
```

```python
plt.figsize = (13,5)
plot_tree(model,filled = True)
plt.show
```

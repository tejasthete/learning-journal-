```python
import pandas as pd
student  = ["A" , "B" , "C" , "D" , "E" , "F" , "G" , "H" , "I" ] 
study_H = [1,2,3,4,5,6,7,8,9] 
Marks = [10,20,30,40,50,60,70,80,90]

data = pd.DataFrame({"Student":student , "Hour" : study_H , "Marks" : Marks })
print(data)
```

```python
x = data[["Hour" , "Marks"]]
print(x)
```

```python
from sklearn.cluster import KMeans
```

```python
model = KMeans(n_clusters = 3 , random_state = 42 )
```

```python
data["Cluster"]=model.fit_predict(x)
print(data)
```

```python
print(model.cluster_centers_)
```

```python
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize= (9,6))
colors = ["Red" , "Green" , "Blue" ]
for i in range(3):
    group = data[data["Cluster"] == i]
    
    plt.scatter(group["Hour"],group["Marks"],color=colors[i],s=100,label=f"Cluster {i}" )
    
    plt.scatter(model.cluster_centers_[:,0],model.cluster_centers_[:,1],s  = 200,color = "black",label = "CLusters",marker = '*')

plt.xlabel("Hour")
plt.ylabel("Marks")
plt.title("K-Means Clustering")
plt.legend()
plt.grid()
plt.show()
```

```python
customer_name = ["C1" , "C2" , "C3" , "C4" , "C5" , "C6" ] 
customer_age= [10,15,20,25,30,35,] 
spending_amt = [5000,10000,15000,20000,25000,30000]

Data = pd.DataFrame({"Customer_Names":customer_name , "Customer_Age" : customer_age , "Monthly_spedning_amt" :spending_amt })
print(Data)
```

```python
X = Data[["Customer_Age" , "Monthly_spedning_amt"]]
print(X)
```

```python
Model = KMeans(n_clusters = 2 , random_state = 42 )
```

```python
Data["Cluster"]=Model.fit_predict(X)
print(Data)
```

```python
print(Model.cluster_centers_)
```

```python
plt.figure(figsize= (9,6))
colors = ["Red" , "Green"  ]
for i in range(2):
    group = Data[Data["Cluster"] == i]
    
    plt.scatter(group["Customer_Age"],group["Monthly_spedning_amt"],color=colors[i],s=100,label=f"Cluster {i}" )
    
    plt.scatter(Model.cluster_centers_[:,0],Model.cluster_centers_[:,1],s  = 200,color = "black",label = "CLuster",marker = '*')

plt.xlabel("Customer_Age")
plt.ylabel("Monthly_spending")
plt.title("K-Means Clustering")
plt.legend()
plt.grid()
plt.show()
```

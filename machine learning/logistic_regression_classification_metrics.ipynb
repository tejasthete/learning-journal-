```python
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import train_test_split
```

```python
x = [[1,50] , [2,55] , [3,60] ,[4,65],[5,70],[6,75],[7,80],[8,85]]
y = [0,0,0,0,1,1,1,1]
```

```python
x_train ,x_test , y_train , y_test  = train_test_split(x,y,test_size = 0.2 ,random_state = 42)
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
x_train = scaler.fit_transform(x_train)
x_test = scaler.transform(x_test)
print(x_train)
print(x_test)
```

```python
print("Orginal Data----->",x_train)

print("Input Data------->",x_test)

print("Original Data---->",y_train)

print("Output Data------>",y_test)
```

```python
model = LogisticRegression()
model.fit(x_train , y_train)
pred = model.predict(x_test)


prediction = model.predict(x_test)
print("Actual_Prediction------>", y_test)
print("Prediction_Data-------->",prediction)
```

```python
from sklearn.metrics import confusion_matrix
cm = confusion_matrix(y_test , pred)
cm
```

```python
from sklearn.metrics import(accuracy_score , precision_score , recall_score , f1_score)
ac = accuracy_score(y_test , pred)
ps = precision_score(y_test , pred)
rs  =recall_score(y_test , pred)
f1 = f1_score(y_test , pred)
print("Accuracy Score----->",ac)
print("Precision Score---->",ps)
print("Recall Score------>",rs)
print("F1 Score---------->",f1)
```

```python
from sklearn.metrics import classification_report
cr = classification_report(y_test , pred)
print("Classification Report",cr)
```

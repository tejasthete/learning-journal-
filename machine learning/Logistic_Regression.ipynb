```python
from sklearn.linear_model import LogisticRegression
```

```python
x = [[1],[2],[3],[4],[5],[6],[7],[8],[9]]
y =[0,0,0,0,0,1,1,1,1]
```

```python
model = LogisticRegression()
model.fit(x,y)
```

```python
predication = model.predict([[10]])
print(predication)
```

```python
res = model.predict_proba([[10]])
print("Probability---->",res)
```

```python
x = [[3],[4],[5],[6],[7],[8]]
y =[0,0,0,1,1,1,]
```

```python
model = LogisticRegression()
model.fit(x,y)
predication = model.predict([[3]])
print(predication)
```

```python
probablity = model.predict_proba([[3]])
print("Probatity---->",probablity)
```

```python
# a=0.94275876
# b=0.05724124
# a+b #-----> probality should always be 1
```

```python
#new predicted
predication1 = model.predict([[10]])
print(predication1)
```

```python
prob1 = model.predict_proba([[10]])
print("Probability---->",prob1)
```

```python
"""
Actual value     Predication data

    1                   1----->True Postive  
    0                   0----->True Negative 
    1                   0----->False Negative 
    0                   1----->False Positive  
"""
```

```python
prob =model.predict_proba(x)
print(prob)
```

```python
#print only class 1
#positive data 
probability1 = []
for i in prob:
    probability1.append(i[1])
print(probability1)
    
    
```

```python
threshold = 0.5
#p<1---> 0
#p>t---->1
predication=[]

for i in probability1:
    if i >= threshold:
        predication.append(1)

    else:
        predication.append(0)
print(predication)
```

```python
print("Actaul_Predication")
print(y)
print()
print("Model_Predication")
print(predication)
```

```python
from sklearn.metrics import confusion_matrix
```

```python
cm = confusion_matrix(y,predication)
print(cm)
```

```python
from sklearn.metrics import accuracy_score
accuracy = accuracy_score(y,predication)
print(accuracy)
```

```python
from sklearn.metrics import precision_score
preci = precision_score(y,predication)
print(preci)
```

```python
#precision = TP/ TP+FP
#recal = TP/TP+
#F1 = 2* precision 
```

```python
from sklearn.metrics import recall_score
rs = recall_score(y,predication)
print(rs)
```

```python
from sklearn.metrics import f1_score
f1 = f1_score(y,predication)
print(f1)
```

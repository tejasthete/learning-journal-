```python
#Label encoder 
#probelm is each element one number will display 
from sklearn.preprocessing import LabelEncoder

data = ["Python" , "Java" , "SQL" , "Python" , "Java" ]

encoder = LabelEncoder()   #Var_name = LabelEncoder()
res = encoder.fit_transform(data)
print(res)


#first collects the unique element and then arranges on the basis of alaphabetical order 

```

```python
#One hot encoder 
#number will display in the form of column 
#each element gets each column 
#only displys two things 1 or 0
#if dtpye is not mention then output will be in form of boolean 
```

```python
import pandas as pd

df = pd.DataFrame({"Lang":data})

print(df)
print()
df= pd.get_dummies(df,columns = ["Lang"],dtype= int)
print(df)
```

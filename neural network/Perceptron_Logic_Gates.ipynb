```python
import numpy as np
```

```python
x = np.array([[0,1],[1,0],[0,0],[1,1]])
x
```

```python
y = np.array([0,0,0,1])
```

```python
from sklearn.linear_model import Perceptron
```

```python
model = Perceptron(max_iter = 1000 , eta0 = 0.1 , tol= None , random_state = 42 , shuffle = True)
```

```python
print(model)
```

```python
model.fit(x,y)
```

```python
prediction = model.predict(x)
```

```python
prediction
```

```python
print("Coeficent",model.coef_)
```

```python
print("Intercept" , model.intercept_)
```

```python
from sklearn.metrics import accuracy_score
```

```python
ascore = accuracy_score(y,prediction)
print("Accuracy Score",ascore)
```

```python
from sklearn.metrics import classification_report
```

```python
report = classification_report(y,prediction)
print("Classification_report",report)
```

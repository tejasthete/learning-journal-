```python
import numpy as np 
```

```python
x = np.array([[0,0],[0,1],[1,0],[1,1]])
```

```python
y = np.array([0,1,1,1])
```

```python
from sklearn.linear_model import Perceptron
```

```python
model = Perceptron(max_iter = 1000 , eta0 = 1 , tol = None , random_state = 42 , shuffle = False)
model
```

```python
model.fit(x,y)
```

```python
predication = model.predict(x)
print(predication)
```

```python
for i in range(len(x)):
    print("Input---->",x[i],
      "Expected->",y[i],
      "Prediction->",predication[i])
```

```python
import matplotlib.pyplot as plt 
x_axis = np.arange(len(x))
x_axis
```

```python
plt.figure(figsize = (8,6))
```

```python
plt.bar(x_axis-0.2 , y, label='Expected' , width = 0.3)
plt.bar(x_axis+0.2 , predication, label='Prediction' , width = 0.3)
plt.legend()
```

```python
z_values = np.linspace(-5,5,200)
z_values
```

```python
step_value = (z_values>=0).astype(int)
step_value
```

```python
plt.step(z_values,step_value , label = "step_value_date")
plt.legend()
plt.grid()
plt.show()
```

```python
sigmod  = 1/1+np.exp(-z_values)
print(sigmod)
```

```python
plt.plot(z_values,sigmod, label = "sigmod")
plt.legend()
plt.show()
```

```python
tanh_value = np.tanh(z_values)
tanh_value
```

```python
plt.plot(z_values,tanh_value , label = "tanh_value")
plt.legend()
plt.grid()
plt.show()
```

```python
relu_value = np.maximum(-1,z_values)
relu_value
```

```python
plt.plot(z_values,relu_value , label = "relu_date")
plt.legend()
plt.grid()
plt.show()
```

```python
softmax= np.exp(z_values)/np.sum(np.exp(z_values))
softmax
```

```python
plt.plot(z_values,softmax , label = "softmax")
plt.legend()
plt.grid()
plt.show()
```
